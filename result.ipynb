```python
import pickle
import os
```

# Result

```python

results_folders = [f for f in os.listdir("results") if os.path.isdir(os.path.join("results", f))]
for folder in results_folders:
    print(f"Processing folder: {folder}")
    for filename in os.listdir(os.path.join("results", folder)):
        file_path = os.path.join("results", folder, filename)
        if os.path.isfile(file_path):
            #print(f"  Found file: {filename}")
            if filename == 'performance.pkl':
                with open(file_path, 'rb') as f:
                    res = pickle.load(f)
                    if filename == 'performance.pkl':
                        print(res['mte'])
                    else:
                        print(res)  
                    print()
```
